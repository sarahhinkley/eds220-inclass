**pandas** is a Python package to wrangle and analyze **tabular data**. Tabular data is data organized into rows and columns, basically like a spreadsheet or a table.

```python
import pandas as pd
import numpy as np
```

A **series** is a one-dimensional array of indexed data. A `pandas.Series` has an index and a `NumPy array` does not have an index. 

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# A series from a numpy array 
pd.Series(np.arange(3), index=[2023, 2024, 2025])
# For this method, we needed to pass a NumPy array (or a list of objects that can be converted to NumPy types) as data
```

```python
# A series from a list of strings with default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

# Initialize series using a dictionary
pd.Series(d)
```

```python
# Creating a pandas.Series from a single value
# If we only provide a single number, boolean, or string as the data for the series we need to provide an index. The value will be repeated to match the length of the index. 
pd.Series(3.0, index = ['A', 'B', 'C'])
```

Arithmetic operations work on series and so do most NumPy functions. 

```python
# Define a series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in series
print(np.exp(s), '\n')

# Original series is unchanged
print(s)
```

```python
# We can also produce new pandas.Series with True/False values indicating whether the elements in a series satisfy a condition or not. 
s > 70
```

```python
# Series with NAs in it
# In pandas we can represent a missing, NULL, or NA value with the float value numpy.nan, which stands for "not a number". 
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs
# The hasnans attribute for a pandas.Series returns True if there are any NA values in it and False otherwise.
s.hasnans
```

```python
# The isna attribute lets us see which elements in the series are NAs.
s.isna()
# The output is a pandas.Series of boolean values indicating if an element in the row at the given index is np.nan (True=is NA) or not (False=not NA)
```

**DATA FRAMES**
The pandas.DataFrame represents tabular data and we can think of it as a spreadsheet. Each column of a pandas.DataFrame is a pandas.Series (it is a dictionary of pandas.Series, with each column name being the key and the column values being the key's value)

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index
df.index = ['a','b','c']
df
```
