```python
import pandas as pd
import os
import matplotlib.pyplot as plt

import statsmodels.formula.api as smf
```

```python
# Read data
snowshoe_hares = pd.read_csv("../data/55_Hare_Data_2012.txt")
```

```python
snowshoe_hares.head()
```

```python
snowshoe_hares.info()
```

```python
# Change the date column from an object to a datetime
snowshoe_hares['date'] = pd.to_datetime(snowshoe_hares['date'])
print(snowshoe_hares['date'].dtype)
```

## Examine the hare's age data

```python
snowshoe_hares['age'].value_counts(dropna=False)
```

When you make a decision, document it. 
    We will keep all hares with a j as juvenile, and all hares with a as an adult. 

```python
# Update age values to juvenile and adult for observations where 'a' and 'j' are included
snowshoe_hares.loc[snowshoe_hares['age'].str.contains('j', case = False, na = False), 'age'] = 'juvenile'
snowshoe_hares.loc[snowshoe_hares['age'].str.contains('a', case = False, na = False), 'age'] = 'adult'

# Filter for confirmed observations of juvenile and adult hares
hares_juv_adult = snowshoe_hares[snowshoe_hares['age'].isin(['juvenile', 'adult'])].copy()
print('Number of juvenile and adult hare observations (captures):', len(hares_juv_adult))
```

```python
# Extract year in which each observation happened
hares_juv_adult['year'] = hares_juv_adult['date'].dt.year
```

```python
# Count observations per year and age, then move age values into columns

hares_per_year = hares_juv_adult.groupby(['year', 'age']).size().unstack()
hares_per_year
```

```python
hares_per_year.index
```

```python
hares_per_year.plot(kind = 'bar',
                    stacked = True,
                    title = 'Juvenile and Adult Snowshoe Hare Captures Per Year',
                    xlabel = 'Year', 
                    ylabel = 'Number of Captures')
```

```python
hares_per_year[['juvenile', 'adult']].plot(kind = 'bar',
                                            stacked = True,
                                            title = 'Juvenile and Adult Snowshoe Hare Captures Per Year',
                                            xlabel = 'Year', 
                                            ylabel = 'Number of Captures')
```

## Plotting with matplotlib

```python
# Initialize an empty figure (fig) and axis (ax)
fig, ax = plt.subplots()

# Add scatter plot of hares data to our figure's axes
hares_juv_adult.plot(kind = 'scatter', x = 'hindft', y = 'weight', ax = ax, alpha = 0.5) 

# Update the axes
ax.set_title('Hind foot length and weight of Snowshoe Hares')
ax.set_xlabel('Hind foot length(mm)')
ax.set_ylabel('Weight (g)')

# Display the figure
plt.show()
```

## Linear Regression 

For our analysis, we want to investigate the possible relations between hind foot length and weight for juvenile hares. 

```python
juv_weight_hindft = hares_juv_adult[hares_juv_adult['age'] == 'juvenile'].dropna(subset = ['hindft', 'weight']) 
# dropna(subset = ['hindft', 'weight'] is saying to drop any rows that have NAs in the hindft or weight columns
```

```python
print('Number of observations:', len(juv_weight_hindft))
juv_weight_hindft.plot.scatter(x = 'hindft', y = 'weight', alpha = 0.5)
```

```python
# Fit model: weight explained by hind foot length
model = smf.ols('weight ~ hindft', data = juv_weight_hindft).fit()
```

```python
# The attribute params extracts the necessary info from model
model.params
```

```python
# Retrieve parameters (save them in two different variables)
intercept = model.params['Intercept']
print('y-intercept:', intercept)

slope = model.params['hindft']
print('slope:', slope)
```

An example of extrapolating beyond what your model can actually capture is interpreting the y-intercept. The y-intercept is saying the weight in grams of the hare when the hind foot length is 0 mm is -352, which does not make sense. 

```python
print('R^2:', model.rsquared)
```

R^2 is the coefficient of determination. It is between 0 and 1. R^2 = 1 is a perfect fit of the model to the data. We interpret it as the proportion of variance in the response that is explained by the least squares line with the predictor variable. 

```python
# Smallest and largest hind foot lengths
x_start = juv_weight_hindft['hindft'].min()
x_end = juv_weight_hindft['hindft'].max()

print(x_start, x_end)
```

```python
# Add the linear regression line into the plot


# Initialize an empty figure (fig) and axis (ax)
fig, ax = plt.subplots()

# Add scatter plot of hares data to our figure's axes
juv_weight_hindft.plot(kind = 'scatter', x = 'hindft', y = 'weight', ax = ax, alpha = 0.5) 

# Add regression line to axes
ax.plot([x_start, x_end], 
        [intercept + slope*x_start, intercept + slope*x_end],
        color = 'darkred')

# Update the axes
ax.set_title('Hind foot length and weight of juvenile Snowshoe Hares')
ax.set_xlabel('Hind foot length(mm)')
ax.set_ylabel('Weight (g)')

# Display the figure
plt.show()
```
