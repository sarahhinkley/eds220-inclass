## Arctic communities by country

We want to calculate the number of Arctic communities by country

````python
# Create a data frame with number of communties per country
n_comms = communities.groupby('country').size().reset_index(name='n_communities')
````

Goal: merge the `n_comms` and the `countries` data frames. 

To merge two data frames they need to have at least one column in common
Do they have a column in common?

````python
# Created a set, a specific type of object in Python
set(countries.columns)

# Tells us if the two dataframes have any columns in common
set(countries.columns).intersection(n_comms.columns)

# If it returns an empty set() we know there are no common columns, a common column would appear in the set()
````

````python
common_columns = set(countries.columns).intersection(n_comms.columns)

# Check if there are any common columns

if len(common_columns) != 0:
    print(f"Common columns: {common_columns}")
else:
    print("No common columns")
````

Check-in: 
Make the previous code into a function called `check_common_columns`, that takes in two data frames as arguments. 

````python
def check_common_columns(df1, df2):
    common_columns = set(df1.columns).intersection(df2.columns)
    
    if len(common_columns) != 0:
        print(f"Common columns: {common_columns}")
    else: 
        print("No common columns")
````

## Merge data frames

We can use the `country_names` data frame to add the country names into the `n_comms`,

To merge data frames we can use `pandas.merge()` function

```python
output_df = pd.merge(left_df,
                   right_df,
                   how = type_of_join,
                   on = column_to_join)
```

where:

- `output_df` = df resulting from merge
- `left_df` = df we have "on the left side"
- `right_df` = df we have "on the right side"
- `how` = type of join
- `on` = column to join on, must be present on both data frames

````python
check_common_columns(n_comms, country_names)
````

````python
n_comms = pd.merge(n_comms, 
                  country_names, 
                  how = 'left',
                  on = 'country')

n_comms
````

````python
check_common_columns(n_comms, countries)
````

Check-in: Create a new variable named `arctic_countries` which is the result of an inner join between our `countries` and `n_comms` data frames. The inner join will merge the subset of rows that have matches in both the left table and the right table. 

````python
# Method 1 to merge data frames 
arctic_countries = pd.merge(countries,
                           n_comms, 
                           how = 'inner',
                           on = 'admin')

# Method 2 to merge data frames
# Here we are already telling it that countries data frame is on the left
arctic_countries_2 = countries.merge(n_comms, how = 'inner', on = 'admin')
arctic_countries_2
````

NOTE: when merging with a GeoDataFrame, the GeoDataFrame needs to be the left data frame to keep the geometry. 

## Review results

Remember that when you are doing a merge, you will often lose some data. 

````python
print(f"Countries in n_comms: {len(n_comms)}")
print(f"Countries in arctic_communities: {len(arctic_countries)}")
````

````python
# Countries in n_comms without a match in countries
n_comms[~n_comms['admin'].isin(countries['admin'])]
````

````python
countries[countries['admin'].str.contains('Aland')]
# From my coutnries data frame, I am looking at the values in my admin column and seeing if any of those contain 'Aland'
````

````python
# Match the name used in the countries data
n_comms['admin'] = n_comms['admin'].replace('Aland Islands', 'Aland')

# Merge again
arctic_countries = countries.merge(n_comms, how = 'inner', on = 'admin')

print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

# Choropleth map 

````python
arctic_countries.plot(column = 'n_communities', # this parameter indicates which column to color by
                     legend = True)
````

## Reprojecting

````python
# Reproject to CRS optimized for Arctic region
arctic_countries = arctic_countries.to_crs('epsg:3413')
````
